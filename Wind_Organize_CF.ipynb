# 2015-2039

```python
#   SSP 126 2015-2039
#create a new column called 'Month' that is the month of the 'time' column
SSP126_2015_2039_tt['Month'] = SSP126_2015_2039_tt['time'].astype(str).str[5:7]
#create a new column called 'Day' that is the day of the 'time' column
SSP126_2015_2039_tt['Day'] = SSP126_2015_2039_tt['time'].astype(str).str[8:10]
#create a new column called 'TimeSlice' that is 'M' + Month + 'D' + Day
SSP126_2015_2039_tt['TimeSlice'] = 'M' + SSP126_2015_2039_tt['Month'] + 'D' + SSP126_2015_2039_tt['Day']
# add a new column called 'Attribute' that is entirely full of 'RNW'
SSP126_2015_2039_tt['Attribute'] = 'RNW'
# add a new column called 'Yr' that is the year of the 'time' column
SSP126_2015_2039_tt['Yr'] = SSP126_2015_2039_tt['time'].astype(str).str[:4]
# add a new column called 'PSET_CI' that is 'ELCWNO' for onshore or 'ELCWNF' for offshore
SSP126_2015_2039_tt['PSET_CI'] = 'ELCWNO'

SSP126_2015_2039_tt= SSP126_2015_2039_tt.groupby(['State/Province', 'time']).first().compute()
#pivot the table so that TEMELE_REGIONS_COPY are the columns, TimeSlice are the rows, and the values are the CP
SSP126_2015_2039_tt = SSP126_2015_2039_tt.pivot_table(index=['TimeSlice', 'Yr', 'Month', 'Day', 'time', 'Attribute', 'PSET_PN'], columns='State/Province', values='CP', aggfunc='first')
#replace all NaN values with 0
SSP126_2015_2039_tt = SSP126_2015_2039_tt.fillna(0)
SSP126_2015_2039_tt
```

```python
#export to csv
SSP126_2015_2039_tt.to_csv('SSP126_2015_2039_TIMES.csv')
```
