Apply IEC Class

```python
#group by loc and find the mean sfc_wind80m_floor. Uncomment and run one at a time. this takes a while to run [~30 mins]


SSP126_2015_2039_iec = SSP126_2015_2039.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2040_2064_iec = SSP126_2040_2064.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2065_2089_iec = SSP126_2065_2089.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2090_2100_iec = SSP126_2090_2100.groupby('loc')['sfc_wind80m_floor'].mean().compute()
```

```python
def IEC_Class(s): # s being sfc_wind80m_foor  25-year avg 
    if s < 7.5:
        return 1
    elif  7.5 <= s <= 8.5:
        return 2
    elif s > 8.5:
        return 3
```

```python
#apply the IEC_Class function to the mean sfc_wind80m_floor in SSP126_2015_2039_iec and name the column 'iec'. Uncomment and run one at a time

SSP126_2015_2039_iec_n = SSP126_2015_2039_iec.apply(IEC_Class).rename('iec')
SSP126_2040_2064_iec_n = SSP126_2040_2064_iec.apply(IEC_Class).rename('iec')
SSP126_2065_2089_iec_n = SSP126_2065_2089_iec.apply(IEC_Class).rename('iec')
SSP126_2090_2100_iec_n = SSP126_2090_2100_iec.apply(IEC_Class).rename('iec')
```

```python
# merge the two dataframes on loc. Uncomment and run one at a time
 
SSP126_2015_2039 = dd.merge(SSP126_2015_2039, SSP126_2015_2039_iec_n, on='loc', how='left')
SSP126_2040_2064 = dd.merge(SSP126_2040_2064, SSP126_2040_2064_iec_n, on='loc', how='left')
SSP126_2065_2089 = dd.merge(SSP126_2065_2089, SSP126_2065_2089_iec_n, on='loc', how='left')
SSP126_2090_2100 = dd.merge(SSP126_2090_2100, SSP126_2090_2100_iec_n, on='loc', how='left')
```

# Organize into TIMES Time Slices 

```python
import polars as pl

# Read all the *TIMES.csv files in Polars and concatenate them
SSP126_TIMES = pl.read_csv('SSP370_2015_2039_TIMES_offshore.csv')
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP370_2040_2064_TIMES_offshore.csv'))
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP370_2065_2089_TIMES_offshore.csv'))
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP370_2090_2100_TIMES_offshore.csv'))

# Filter the DataFrame to keep only rows with 'Yr' greater than or equal to 2020
SSP126_TIMES = SSP126_TIMES.filter(pl.col('Yr') >= 2020)
```

```python
SSP126_TIMES.head()
```

```python
import polars as pl

# Add 'Season' column based on the 'Month' column
SSP126_TIMES = SSP126_TIMES.with_columns(
    Season=
    SSP126_TIMES["Month"].apply(
        lambda month: "S01" if month in [1, 2, 12] else "S02" if month in [3, 4, 5] else "S03" if month in [6, 7, 8] else "S04"
    ),
)

#read 'time' column as datetime 

SSP126_TIMES = SSP126_TIMES.with_columns(pl.col('time').str.strptime(pl.Datetime).alias('time'))

# find the weekday number of the 'time' column 
SSP126_TIMES = SSP126_TIMES.with_columns(
    Weekday=
    SSP126_TIMES["time"].apply(
        lambda time: time.isoweekday()
    ),
)

# if the 'Weekday' is equal to 6, 2, 3, 4, or 5, then the 'Type' is 'WD2', otherwise it is 'WD1'
SSP126_TIMES = SSP126_TIMES.with_columns(
    Type=
    SSP126_TIMES["Weekday"].apply(
        lambda Weekday: "WD2" if Weekday in [6, 2, 3, 4, 5] else "WD1"
    ),
)

SSP126_TIMES = SSP126_TIMES.with_columns(pl.lit('NCAP_AF').alias('Attribute'))
```

```python
# Group by year, month, type, then calculate the average for each groupby 
#groupby the Yr and Season and find mean 
SSP126_TIMES_120 = SSP126_TIMES.groupby(['Yr', 'Month', 'Type']).mean()
# sort by year and season
SSP126_TIMES_120 = SSP126_TIMES_120.sort(['Yr', 'Month'])
# make all rows in the 'PSET_CI' column 'ELCWNF' in polars
# add a new column called 'PSET_CI' that is 'ELCWNO' for onshore or 'ELCWNF' for offshore
SSP126_TIMES_120 = SSP126_TIMES_120.with_columns(pl.col('PSET_CI').fill_null('ELCWNF'))
# do the same but RNW for Attribute
SSP126_TIMES_120 = SSP126_TIMES_120.with_columns(pl.col('Attribute').fill_null('NCAP_AF'))
# make a column called 'M' that is just 'M' + Month
SSP126_TIMES_120 = SSP126_TIMES_120.with_columns(pl.lit('M').alias('M'))
#create a new column called 'TimeSlice' that is 'M' + Month + Type
SSP126_TIMES_120 = SSP126_TIMES_120.with_columns(pl.concat_str([pl.col('M'), pl.col('Month').cast(pl.Utf8).str.zfill(2), pl.col('Type')], separator='').alias('TimeSlice'))
# drop TimeSlice, time, Attribute, and PSET_CI columns
SSP126_TIMES_120 = SSP126_TIMES_120.drop(['Day', 'time', 'Type', 'M', 'Season', 'Weekday', 'Month'])

SSP126_TIMES_120
```

# TS - 40
Averaged by season and weekday type by country. ex: Weekdays in Fall in 2020 in USA, Weekends in Fall in 2020 in USA

```python
#groupby the Yr and Season and find mean 
SSP126_TIMES_40 = SSP126_TIMES.groupby(['Yr', 'Season', 'Type']).mean()
# sort by year and season
SSP126_TIMES_40 = SSP126_TIMES_40.sort(['Yr', 'Season'])
# drop TimeSlice, time, Attribute, and PSET_CI columns
SSP126_TIMES_40 = SSP126_TIMES_40.drop(['TimeSlice', 'time', 'Weekday', 'Month', 'Day'])
# make all rows in the 'PSET_CI' column 'ELCWNF' in polars
# add a new column called 'PSET_CI' that is 'ELCWNO' for onshore or 'ELCWNF' for offshore
SSP126_TIMES_40 = SSP126_TIMES_40.with_columns(pl.col('PSET_CI').fill_null('ELCWNF'))
# do the same but RNW for Attribute
SSP126_TIMES_40 = SSP126_TIMES_40.with_columns(pl.col('Attribute').fill_null('NCAP_AF'))
#Create a column called 'TimeSlice' that is the Season + Type
SSP126_TIMES_40 = SSP126_TIMES_40.with_columns(pl.concat_str([pl.col('Season'), pl.col('Type')], separator='').alias('TimeSlice'))
# drop the Season and Type columns
SSP126_TIMES_40 = SSP126_TIMES_40.drop(['Season', 'Type'])

SSP126_TIMES_40
```
