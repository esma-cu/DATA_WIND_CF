```python
#group by loc and find the mean sfc_wind80m_floor. Uncomment and run one at a time. this takes a long time


SSP126_2015_2039_iec = SSP126_2015_2039.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2040_2064_iec = SSP126_2040_2064.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2065_2089_iec = SSP126_2065_2089.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2090_2100_iec = SSP126_2090_2100.groupby('loc')['sfc_wind80m_floor'].mean().compute()
```

```python
def IEC_Class(s): # s being sfc_wind80m_foor  25-year avg 
    if s < 7.5:
        return 1
    elif  7.5 <= s <= 8.5:
        return 2
    elif s > 8.5:
        return 3
```

```python
#apply the IEC_Class function to the mean sfc_wind80m_floor in SSP126_2015_2039_iec and name the column 'iec'. Uncomment and run one at a time

SSP126_2015_2039_iec_n = SSP126_2015_2039_iec.apply(IEC_Class).rename('iec')
SSP126_2040_2064_iec_n = SSP126_2040_2064_iec.apply(IEC_Class).rename('iec')
SSP126_2065_2089_iec_n = SSP126_2065_2089_iec.apply(IEC_Class).rename('iec')
SSP126_2090_2100_iec_n = SSP126_2090_2100_iec.apply(IEC_Class).rename('iec')
```

```python
import holoviews as hv
import pandas as pd

# Assuming your DataFrame is named df
# Extract the decade from the 'time' column
GFDL_offshore['Decade'] = GFDL_offshore['time'].dt.year // 10 * 10

# Define the batch size (e.g., 1000)
batch_size = 1000

# Create an empty list to store plots
plots = []

# Loop through the data in batches
for decade, data_decade in GFDL_offshore.groupby('Decade'):
    for scenario, data_scenario in data_decade.groupby('Scenario'):
        for start in range(0, len(data_scenario), batch_size):
            end = start + batch_size
            batch_data = data_scenario.iloc[start:end]
            box_plot = batch_data.hvplot.box(
                y='Difference_N_m_Hist', by='Country',
                title=f'Decade {decade} - {scenario}', rot=45
            )
            plots.append(box_plot)

# Create an overlay of the box and whisker plots
overlay = hv.Overlay(plots)

# Create a layout with subplots for each plot
layout = hv.NdLayout(overlay).cols(2)  # Adjust the number of columns as needed

# Display the layout
layout.opts(hv.opts.BoxWhisker(width=300, height=300))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap

def plot_decade_average_subplots_with_borders(data):
    # Group the data by 'lat', 'lon', and 'Decade' and calculate the mean for 'Difference_N_m_Hist'
    grouped_data = data.groupby(['lat', 'lon', 'Decade'])['Difference_N_m_Hist'].mean().reset_index()

    # Get unique decades
    decades = sorted(grouped_data['Decade'].unique())

    # Create a figure with increased height
    fig = plt.figure(figsize=(10, 6 * len(decades)))

    # Create a separate subplot for each decade
    for i, decade in enumerate(decades):
        plt.subplot(len(decades), 1, i+1)
        decade_data = grouped_data[grouped_data['Decade'] == decade]
        
        # Create a Basemap instance for the current subplot
        map = Basemap(projection='merc', llcrnrlat=decade_data['lat'].min(), urcrnrlat=decade_data['lat'].max(),
                      llcrnrlon=decade_data['lon'].min(), urcrnrlon=decade_data['lon'].max(), resolution='i')

        # Draw country borders
        map.drawcountries()
        
        # Create a scatter plot for the current decade
        sc = map.scatter(decade_data['lon'], decade_data['lat'], c=decade_data['Difference_N_m_Hist'],
                         cmap='viridis', s=50, marker='o', edgecolor='k')
        
        # Customize subplot labels and title
        plt.title(f'Decade {decade} Average Difference_N_m_Hist')
        plt.xlabel('Longitude')
        plt.ylabel('Latitude')
        plt.colorbar(sc, label='Average Difference_N_m_Hist')

    # Adjust subplot layout and spacing
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.5)  # Adjust the vertical spacing

    # Show the plot
    plt.show()


# Call the function to plot decade averages on subplots
plot_decade_average_subplots(SSP370_offshore)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap

def plot_seasonal_average_subplots_with_borders(data):
    # Define the seasons (you can customize this as needed)
    seasons = {
        'Winter': [12, 1, 2],   # December, January, February
        'Spring': [3, 4, 5],    # March, April, May
        'Summer': [6, 7, 8],    # June, July, August
        'Autumn': [9, 10, 11]   # September, October, November
    }

    # Create a figure with increased height
    fig = plt.figure(figsize=(10, 6 * len(seasons)))

    # Create a separate subplot for each season
    for i, (season_name, months) in enumerate(seasons.items()):
        plt.subplot(len(seasons), 1, i+1)
        
        # Filter the data for the current season
        season_data = data[data['time'].dt.month.isin(months)]
        
        # Create a Basemap instance for the current subplot
        map = Basemap(projection='merc', llcrnrlat=season_data['lat'].min(), urcrnrlat=season_data['lat'].max(),
                      llcrnrlon=season_data['lon'].min(), urcrnrlon=season_data['lon'].max(), resolution='i')

        # Draw country borders
        map.drawcountries()
        
        # Group the data by 'lat', 'lon', and 'Decade' and calculate the mean for 'Difference_N_m_Hist'
        grouped_data = season_data.groupby(['lat', 'lon', 'Decade'])['Difference_N_m_Hist'].mean().reset_index()
        
        # Create a scatter plot for the current season
        sc = map.scatter(grouped_data['lon'], grouped_data['lat'], c=grouped_data['Difference_N_m_Hist'],
                         cmap='viridis', s=50, marker='o', edgecolor='k')
        
        # Customize subplot labels and title
        plt.title(f'{season_name} Average Difference_N_m_Hist')
        plt.xlabel('Longitude')
        plt.ylabel('Latitude')
        plt.colorbar(sc, label='Average Difference_N_m_Hist')

    # Adjust subplot layout and spacing
    plt.tight_layout()
    plt.subplots_adjust(hspace=0.5)  # Adjust the vertical spacing

    # Show the plot
    plt.show()


# Call the function to plot decade averages on subplots
plot_decade_average_subplots(SSP370_offshore)
```
