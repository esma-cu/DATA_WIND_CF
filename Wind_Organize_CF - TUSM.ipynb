```python
# split the 'SSP126' into datasets with the following years: 2015-2039, 2040-2064, 2065-2089, 2090-2100
SSP126_2015_2039 = SSP126[SSP126['Year'] <= 2039]
SSP126_2040_2064 = SSP126[(SSP126['Year'] >= 2040) & (SSP126['Year'] <= 2064)]
SSP126_2065_2089 = SSP126[(SSP126['Year'] >= 2065) & (SSP126['Year'] <= 2089)]
SSP126_2090_2100 = SSP126[SSP126['Year'] >= 2090]
```

Apply IEC Class

```python
#group by loc and find the mean sfc_wind80m_floor. Uncomment and run one at a time. this takes a long time


SSP126_2015_2039_iec = SSP126_2015_2039.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2040_2064_iec = SSP126_2040_2064.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2065_2089_iec = SSP126_2065_2089.groupby('loc')['sfc_wind80m_floor'].mean().compute()
SSP126_2090_2100_iec = SSP126_2090_2100.groupby('loc')['sfc_wind80m_floor'].mean().compute()
```

```python
def IEC_Class(s): # s being sfc_wind80m_foor  25-year avg 
    if s < 7.5:
        return 1
    elif  7.5 <= s <= 8.5:
        return 2
    elif s > 8.5:
        return 3
```

```python
#apply the IEC_Class function to the mean sfc_wind80m_floor in SSP126_2015_2039_iec and name the column 'iec'. Uncomment and run one at a time

SSP126_2015_2039_iec_n = SSP126_2015_2039_iec.apply(IEC_Class).rename('iec')
SSP126_2040_2064_iec_n = SSP126_2040_2064_iec.apply(IEC_Class).rename('iec')
SSP126_2065_2089_iec_n = SSP126_2065_2089_iec.apply(IEC_Class).rename('iec')
SSP126_2090_2100_iec_n = SSP126_2090_2100_iec.apply(IEC_Class).rename('iec')
```

```python
# merge the two dataframes on loc. Uncomment and run one at a time
 
SSP126_2015_2039 = dd.merge(SSP126_2015_2039, SSP126_2015_2039_iec_n, on='loc', how='left')
SSP126_2040_2064 = dd.merge(SSP126_2040_2064, SSP126_2040_2064_iec_n, on='loc', how='left')
SSP126_2065_2089 = dd.merge(SSP126_2065_2089, SSP126_2065_2089_iec_n, on='loc', how='left')
SSP126_2090_2100 = dd.merge(SSP126_2090_2100, SSP126_2090_2100_iec_n, on='loc', how='left')
```

# Organize into TIMES Time Slices 

```python
import polars as pl

# Read all the *TIMES.csv files in Polars and concatenate them
SSP126_TIMES = pl.read_csv('SSP126_projectedstatus_usa_onshore_2015_2039_TIMES_state.csv')
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP126_projectedstatus_usa_onshore_2040_2064_TIMES_state.csv'))
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP126_projectedstatus_usa_onshore_2065_2089_TIMES_state.csv'))
SSP126_TIMES = SSP126_TIMES.extend(pl.read_csv('SSP126_projectedstatus_usa_onshore_2090_2100_TIMES_state.csv'))

# Filter the DataFrame to keep only rows with 'Yr' greater than or equal to 2020
SSP126_TIMES = SSP126_TIMES.filter(pl.col('Yr') >= 2020)
```

```python
SSP126_TIMES.head()
```

```python
import polars as pl

# Add 'Season' column based on the 'Month' column
SSP126_TIMES = SSP126_TIMES.with_columns(
    Season=
    SSP126_TIMES["Month"].apply(
        lambda month: "S01" if month in [1, 2, 12] else "S02" if month in [3, 4, 5] else "S03" if month in [6, 7, 8] else "S04"
    ),
)

#read 'time' column as datetime 

SSP126_TIMES = SSP126_TIMES.with_columns(pl.col('time').str.strptime(pl.Datetime).alias('time'))

# find the weekday number of the 'time' column 
SSP126_TIMES = SSP126_TIMES.with_columns(
    Weekday=
    SSP126_TIMES["time"].apply(
        lambda time: time.isoweekday()
    ),
)

# if the 'Weekday' is equal to 6, 2, 3, 4, or 5, then the 'Type' is 'WD2', otherwise it is 'WD1'
SSP126_TIMES = SSP126_TIMES.with_columns(
    Type=
    SSP126_TIMES["Weekday"].apply(
        lambda Weekday: "WD2" if Weekday in [6, 2, 3, 4, 5] else "WD1"
    ),
)

SSP126_TIMES = SSP126_TIMES.with_columns(pl.lit('NCAP_AF').alias('Attribute'))
```

8760 Method for COM_FR

This is the same as earlier in the code, however, there is no timeslice weighting, so divide the CF by the annual sum

```python
import pandas as pd
test = pd.read_excel('SSP126_projectedstatus_usa_offshore_TS8760_TUSM.xlsx')
test
```

```python
# List of US_* column names
us_columns = [col for col in test.columns if col.startswith('US_')]

# rename us_columns by adding _n to the end of each column name in us_columns. remove the us_columns from test
test = test.rename(columns={col: col + '_n' for col in us_columns})
test
```

```python
# group by Year and find the sum of each column, rename each column with '_sum' at the end
test_sum = test.groupby('Yr').sum().add_suffix('_sum').reset_index()
```

```python
# merge the two on the 'Yr' column
test = pd.merge(test, test_sum, on='Yr', how='left')
test.head()
```

```python
# if the column name ends with '_n', then divide the column by the column with the same name but with '_sum' at the end
for col in test.columns:
    if col.endswith('_n'):
        test[col] = test[col] / test[col + '_sum']
test
```

```python
# drop all columns that end with '_sum'
test = test.drop(test.filter(regex='_sum$').columns, axis=1)
# remove the '_n' from the column names
test.columns = test.columns.str.replace('_n', '')
# Replace the values in Attribute with COM_FR
test['Attribute'] = 'COM_FR'
# Replace the column name PSet_CI = Cset_CN 
test = test.rename(columns={'PSet_CI': 'CSet_CN'})
```

```python
# drop Month, Day, time, Season, Weekday, and Type columns
test = test.drop(['Month', 'Day', 'time', 'Season', 'Weekday', 'Type'], axis=1)
```
