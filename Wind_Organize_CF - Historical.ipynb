```python
# 25 MB chunks 

historical = dd.read_csv('historical_usa_onshore_daily_1981_2014.csv', blocksize=25e6, dtype={'lat': 'float64','lon': 'float64','sfcwind': 'float64','Capacity (MW)': 'float64', 'City' : 'string','Local area (taluk, county)' :'string','Major area (prefecture, district)':'string', 'State/Province':'string', 'Region':'string','Country':'string','sfc_wind80m':'float64','sfc_wind80m_floor':'float64'})
```

```python
historical['Year'] = historical['time'].astype(str).str[:4] # creating year column
historical['Year'] = historical['Year'].astype(int) 
# add a column that is called 'loc' and it is the combination of lat and lon with a comma in between
historical['loc'] = historical['lat'].astype(str) + ',' + historical['lon'].astype(str)


historical.head()
```

Apply IEC Class

```python
#group by loc and find the mean sfc_wind80m_floor. Uncomment and run one at a time. this takes a long time
historical_iec = historical.groupby('loc')['sfc_wind80m_floor'].mean().compute()
```

```python
historical_iec
```

```python
def IEC_Class(s): # s being sfc_wind80m_foor  25-year avg 
    if s < 7.5:
        return 1
    elif  7.5 <= s <= 8.5:
        return 2
    elif s > 8.5:
        return 3
```

```python
#apply the IEC_Class function to the mean sfc_wind80m_floor in SSP126_2015_2039_iec and name the column 'iec'. Uncomment and run one at a time

historical_iec_n = historical_iec.apply(IEC_Class).rename('iec')
```

```python
# merge the two dataframes on loc. Uncomment and run one at a time
 
historical = dd.merge(historical, historical_iec_n, on='loc', how='left')
```

```python
#rename 'sfc_wind80m_floor_x' to 'sfc_wind80m_floor'. Uncomment and run one at a time
historical = historical.rename(columns={'sfc_wind80m_floor_x': 'sfc_wind80m_floor'})
```

```python
IEC_Classifications = dd.read_csv('iec_classifications.csv')
# convert IEC_Class to i64
IEC_Classifications.compute()
```

Now we apply the Capacity Factors to each unique IEC Class

```python
# merge the data on 'iec' and 'sfc_wind80m'. Uncomment and run one at a time
historical_n = dd.merge(historical, IEC_Classifications, on=['iec', 'sfc_wind80m_floor'], how='left')
```

```python
# average the CP by TEMELE_REGIONS_COPY and time. Uncomment and run one at a time. this takes a long time
historical_tt = historical_n.groupby(['State/Province', 'time'])['CP', 'Capacity (MW)'].mean().compute()
```

Organize in the Format that is compliant with TIMES/VEDA 2.0 --> Only Run the Block associated with the proper SSP

```python
# dask open csv
historical_tt = dd.read_csv('historical_usa_onshore_test_state.csv')
```

```python
#create a new column called 'Month' that is the month of the 'time' column
historical_tt['Month'] = historical_tt['time'].astype(str).str[5:7]
#create a new column called 'Day' that is the day of the 'time' column
historical_tt['Day'] = historical_tt['time'].astype(str).str[8:10]
#create a new column called 'TimeSlice' that is 'M' + Month + 'D' + Day
historical_tt['TimeSlice'] = 'M' + historical_tt['Month'] + 'D' + historical_tt['Day']
# add a new column called 'Attribute' that is entirely full of 'RNW'
historical_tt['Attribute'] = 'NCAP_AF'
# add a new column called 'Yr' that is the year of the 'time' column
historical_tt['Yr'] = historical_tt['time'].astype(str).str[:4]
# add a new column called 'PSET_CI' that is 'ELCWNO' for onshore or 'ELCWNF' for offshore
historical_tt['PSET_CI'] = 'ELCWNO'

historical_tt= historical_tt.groupby(['State/Province', 'time']).first().compute()
#pivot the table so that TEMELE_REGIONS_COPY are the columns, TimeSlice are the rows, and the values are the CP
historical_tt = historical_tt.pivot_table(index=['TimeSlice', 'Yr', 'Month', 'Day', 'time', 'Attribute', 'PSET_CI'], columns='State/Province', values='CP', aggfunc='first')
#replace all NaN values with 0
historical_tt = historical_tt.fillna(0)
# only keep TimeSlice', 'Yr', 'Month', 'Day', 'time', 'Attribute', 'PSET_CI', 'Virginia', 'Rhode Island'
#historical_tt = historical_tt[['Virginia', 'Rhode Island']]
historical_tt
```

```python
#export to csv
historical_tt.to_csv('historical_usa_onshore_TIMES_state.csv')
```

# Organize into TIMES Time Slices 

```python
import polars as pl

# Read all the *TIMES.csv files in Polars and concatenate them
historical_TIMES = pl.read_csv('historical_usa_offshore_TIMES_state.csv')
```

```python
historical_TIMES.head()
```

```python
import polars as pl

# Add 'Season' column based on the 'Month' column
historical_TIMES = historical_TIMES.with_columns(
    Season=
    historical_TIMES["Month"].apply(
        lambda month: "S01" if month in [1, 2, 12] else "S02" if month in [3, 4, 5] else "S03" if month in [6, 7, 8] else "S04"
    ),
)

#read 'time' column as datetime 

historical_TIMES = historical_TIMES.with_columns(pl.col('time').str.strptime(pl.Datetime).alias('time'))

# find the weekday number of the 'time' column 
historical_TIMES = historical_TIMES.with_columns(
    Weekday=
    historical_TIMES["time"].apply(
        lambda time: time.isoweekday()
    ),
)

# if the 'Weekday' is equal to 6, 2, 3, 4, or 5, then the 'Type' is 'WD2', otherwise it is 'WD1'
historical_TIMES = historical_TIMES.with_columns(
    Type=
    historical_TIMES["Weekday"].apply(
        lambda Weekday: "WD2" if Weekday in [6, 2, 3, 4, 5] else "WD1"
    ),
)

historical_TIMES = historical_TIMES.with_columns(pl.lit('NCAP_AF').alias('Attribute'))
```

# TS - 120
Averaged by month and weekday type by country. ex: Weekdays in January in 2020 in USA, Weekends in January in 2020 in USA

```python
# Group by year, month, type, then calculate the average for each groupby 
#groupby the Yr and Season and find mean 
historical_TIMES = historical_TIMES.groupby(['Yr', 'Month', 'Type']).mean()
# sort by year and season
historical_TIMES = historical_TIMES.sort(['Yr', 'Month'])
# make all rows in the 'PSET_CI' column 'ELCWNF' in polars
# add a new column called 'PSET_CI' that is 'ELCWNO' for onshore or 'ELCWNF' for offshore
historical_TIMES = historical_TIMES.with_columns(pl.col('PSET_CI').fill_null('ELCWNF'))
# do the same but RNW for Attribute
historical_TIMES = historical_TIMES.with_columns(pl.col('Attribute').fill_null('NCAP_AF'))
# make a column called 'M' that is just 'M' + Month
historical_TIMES = historical_TIMES.with_columns(pl.lit('M').alias('M'))
#create a new column called 'TimeSlice' that is 'M' + Month + Type
historical_TIMES = historical_TIMES.with_columns(pl.concat_str([pl.col('M'), pl.col('Month').cast(pl.Utf8).str.zfill(2), pl.col('Type')], separator='').alias('TimeSlice'))
# drop TimeSlice, time, Attribute, and PSET_CI columns
historical_TIMES = historical_TIMES.drop(['Day', 'time', 'Type', 'M', 'Season', 'Weekday', 'Month'])

historical_TIMES
```

```python
# save to csv 
#SSP126_TIMES_120.write_csv('SSP585_Onshore_TS120_TUSM.csv') # Onshore
historical_TIMES.write_csv('Historical_Offshore_TS120_TUSM.csv') # Offshore
```
